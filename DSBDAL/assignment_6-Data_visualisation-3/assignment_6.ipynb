```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
df = pd.read_csv("Iris.csv")

# Quick scan
df.head()
```

```python
# Histograms (skip Id and Species)
df.drop(columns=["Id", "Species"]).hist(
    figsize=(10, 8),
    bins=20
)

plt.suptitle("Histograms of Iris Features")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.drop(columns=["Id", "Species"]))
plt.title("Box Plots of Iris Features")
plt.show()
```
