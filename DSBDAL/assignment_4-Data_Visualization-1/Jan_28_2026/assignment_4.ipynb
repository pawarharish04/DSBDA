```python
# import seaborn for dataset access and plotting
import seaborn as sns

# import matplotlib for showing the plots
import matplotlib.pyplot as plt
```

```python
# load the built-in titanic dataset into a DataFrame
titanic = sns.load_dataset('titanic')
```

```python
# set a clean seaborn style
sns.set_style('whitegrid')
```

```python
# drop 'deck' because half the values are missing anyway
titanic = titanic.drop(columns=['deck'])
```

```python
# fill missing 'embarked' with the most common category
titanic['embarked'] = titanic['embarked'].fillna(titanic['embarked'].mode()[0])
```

```python
# fill missing 'age' with the median age
titanic['age'] = titanic['age'].fillna(titanic['age'].median())
```

```python
# check results
titanic.head()
```

```python
titanic.info()
```

```python
# Add index as a column for plotting
titanic["index_col"] = titanic.index

# Line plot of age over index
sns.lineplot(data=titanic, x="index_col", y="age")
plt.title("Line Plot: Passenger Age over Index")
plt.show()
```
