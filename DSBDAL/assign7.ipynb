```python
import nltk
from nltk import word_tokenize, sent_tokenize

nltk.download('punkt')

para = "Sunset is the time of day when our sky meets the outer space solar winds. There are blue, pink, and purple swirls, spinning and twisting, like clouds of balloons caught in a whirlwind. The sun moves slowly to hide behind the line of horizon, while the moon races to take its place in prominence atop the night sky."

words = word_tokenize(para)
sentences = sent_tokenize(para)

print("Sentences:", sentences)
print("Words:", words)
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords 
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.lower() not in stop_words]
print("Filtered Words:", filtered_words)
```

```python
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger_eng')

pos_tags = pos_tag(filtered_words)
print(pos_tags)
```

```python
from nltk.stem import LancasterStemmer
stemmer = LancasterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

def get_wordnet_tag(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('R'):
        return 'r'
    elif tag.startswith('V'):
        return 'v'
    else:
        return 'n'
    
lemmatized_words = [lemmatizer.lemmatize(word, get_wordnet_tag(tag)) for word,tag in pos_tags]

print(lemmatized_words)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

corpus = sentences

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)

tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())

print("TF-IDF Matrix:")
display(tfidf_df)

top_words = tfidf_df.sum().sort_values(ascending=False)
display(top_words.to_frame(name="score"))
```
