```python
mean_SepalLengthCm = df["SepalLengthCm"].mean()                   
#continuous normally distributed numerical variables with a very small proportion of missing values

df["SepalLengthCm"].fillna(mean_SepalLengthCm , inplace=True)     
#checks values in succession if null, and if true, repace the null value with mean_SepalLengthCm

df.info()                                                         
#provides information about the dataset such as missing values and datatypes
```

```python
mean_SepalWidthCm = df["SepalWidthCm"].mean()                    
#continuous normally distributed numerical variables with a very small proportion of missing values

df["SepalWidthCm"].fillna(mean_SepalWidthCm , inplace=True)      
#checks values in succession if null, and if true, repace the null value with mean_SepalWidthCm

df.info()                                                        
#provides information about the dataset such as missing values and datatypes
```
