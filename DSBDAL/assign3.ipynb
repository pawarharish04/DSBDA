```python
import pandas as pd

df = pd.read_csv('loan_data.csv')
df.head()
```

```python
df.describe()
```

```python
bins = [18,35,60,100,150]
labels =['young','middle-aged','old','super-old']

df['age'] = pd.cut(df['person_age'], bins=bins, labels=labels)
display(df.head())
```

```python
display(df['person_income'].groupby(df['age'], observed=False).describe())
```

```python
age_with_val = pd.DataFrame(data=df, columns = ['age','person_age'])
display(age_with_val)
```
