```python
bins = [20, 34, 50, float('inf')]
labels = [1, 2, 3]

df1['age_group'] = pd.cut(df1['Age'], bins=bins, labels=labels, right=True, include_lowest=True)
```

```python
grouped = df1.groupby('age_group')['Salary'].agg(['mean', 'median', 'min', 'max', 'std'])
print(grouped)
```
