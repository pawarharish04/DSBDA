```python
import matplotlib.pyplot as plt

df.hist(column=['SepalLengthCm','SepalWidthCm','PetalLengthCm', 'PetalWidthCm'],edgecolor='yellow',figsize=(10,8))
```

```python
df.plot(column=['SepalLengthCm','SepalWidthCm','PetalLengthCm', 'PetalWidthCm'],kind='box', subplots=True, layout=(2, 3), figsize=(10, 8))
```

```python
import numpy as np

q1 = np.percentile(df['SepalWidthCm'], 25)
q3 = np.percentile(df['SepalWidthCm'], 75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df['SepalWidthCm'] < lower_bound) | (df['SepalWidthCm'] > upper_bound)]

print(f'Lower bound: {lower_bound}')
print(f'Upper bound: {upper_bound}')
print(f'Outlier values: {outliers["SepalWidthCm"].tolist()}')
```

Observation: Only sepal-width has outliers. Petal length has the highest spread as compared to others.
Petal length/width have a clear separation b/w sentosa and others.
