```python
import nltk
import math
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk import pos_tag
from collections import Counter

```

```python
with open("document1.txt","r") as f1:
    t1=f1.read()
with open("document2.txt","r") as f2:
    t2=f2.read()
docs=[t1,t2]
```

```python
# token
token=word_tokenize(t1)
print(token)
```

```python
# pos
pos=pos_tag(token)
print(pos)
```

```python
# stop word
sw=set(stopwords.words("english"))
l=[]
for w in token:
    if w.lower() not in sw:
        l.append(w)
print(l)
```

```python
# stem
st=PorterStemmer()
l1=[]
for w in token:
    l1.append(st.stem(w))
print(l1)
```

```python
# lem
l2=[]
l=WordNetLemmatizer()
for w in token:
    l2.append(l.lemmatize(w))
print(l2)
```

```python
# tf
def fun(x):
    w=x.split()
    n=len(w)
    l={}
    for wd in w:
        l[wd]=w.count(wd)/n
    return l
d1=fun(t1)
d2=fun(t1)
print(d1)
print(d2)
```

```python
# idf
n=len(docs)
idf={}
for d in docs:
    for w in set(d.split()):
        idf[w]=idf.get(w,0)+1
for w in idf:
    idf[w]=math.log(n/idf[w])
print(idf)
```

```python
# tf-idf
l1={}
l2={}
for d in d1:
    if d in idf:
        l1[d]=d1[d]*idf[d]
for d in d2:
    if d in idf:
        l2[d]=d2[d]*idf[d]
print(l1)
print(l2)
```
