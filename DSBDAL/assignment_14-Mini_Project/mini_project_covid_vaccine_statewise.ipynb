# Mini Project - covid_vaccine_statewise

```python
import pandas as pd
```

```python
df = pd.read_csv("covid_vaccine_statewise.csv")
```

```python
# Preview
print("First 5 rows:")
print(df.head())
```

```python
# a) Describe dataset
print("\nDataset Info:")
print(df.info())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Clean column names
df.columns = df.columns.str.strip()
```

```python
# Convert numeric columns
numeric_cols = [
    'First Dose Administered',
    'Second Dose Administered',
    'Male (Doses Administered)',
    'Female (Doses Administered)'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Remove rows where State = 'India'
state_df = df[df['State'] != 'India']
```

```python
# b) First dose state-wise
print("\nFirst Dose (State-wise):")
first_dose = state_df.groupby('State')['First Dose Administered'].max().astype(int)
print(first_dose)
```

```python
# c) Second dose state-wise
print("\nSecond Dose (State-wise):")
second_dose = state_df.groupby('State')['Second Dose Administered'].max().astype(int)
print(second_dose)
```

```python
# d) Total males vaccinated
total_males = df['Male (Doses Administered)'].max().astype(int)
print("\nTotal Males Vaccinated:", total_males)
```

```python
# e) Total females vaccinated
total_females = df['Female (Doses Administered)'].max().astype(int)
print("\nTotal Females Vaccinated:", total_females)
```
