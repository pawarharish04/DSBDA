```python

# Step 2 - Create Age Groups and List
# drop missing values from Age and Salary
df = df.dropna(subset=['Age', 'Salary'])

# create age groups using pd.cut()
bins = [0, 20, 25, 30, 35, 40]
labels = ['0-20', '21-25', '26-30', '31-35', '36-40']


df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=True)
```

```python
# create a list of salaries for each age group
age_group_list = df.groupby('AgeGroup', observed=True)['Salary'].apply(list)

print("\nList of Salaries for each Age Group:")
print(age_group_list)
```

```python
# Step 3 - Summary Statistics Grouped by Age Group
summary = df.groupby('AgeGroup', observed=True)['Salary'].agg(
    Mean='mean',
    Median='median',
    Minimum='min',
    Maximum='max',
    Std_Dev='std'
)

print("\nSummary Statistics of Salary grouped by Age Group:")
print(summary)
```

```python

# =============================================
# Part 2 - Iris Dataset Statistics
# =============================================

# Step 4 - Load Iris Dataset
iris = pd.read_csv("Iris.csv")

print("\nFirst 5 rows:")
print(iris.head())
```

```python
print("\nShape:", iris.shape)

print("\nData Types:")
print(iris.dtypes)

print("\nMissing Values:")
print(iris.isnull().sum())
```

```python
# Step 5 - Summary Statistics for Each Species
def species_stats(species_name):
    # filter data for that species
    data = iris[iris['Species'] == species_name]
    
    print(f"\nStatistics for {species_name}:")
    print(data.describe().transpose())

# call function for each species
species_stats('Iris-setosa')
species_stats('Iris-versicolor')
species_stats('Iris-virginica')
```
